```python
mystring = 'hello'
```

```python
mylist = []

for letter in mystring:
    mylist.append(letter)
```

```python
mylist
```

```python
mylist = [letter for letter in mystring]
mylist
```

```python
mylist = [x for x in 'word']
mylist
```

```python
mylist = [num**2 for num in range(0,11)]
mylist
```

```python
mylist = [num**2 for num in range(0,11) if num%2==0]
mylist
```

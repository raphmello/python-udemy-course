```python
# This is a string: Hello World
print("This is a string: {}".format(x))
```

```python
# Hello World 3
print("{} {} {}".format("Hello", "World", 3))
```

```python
# 3 World Hello
print("{2} {1} {0}".format("Hello", "World", 3))
```

```python
# Hello Hello Hello
print("{0} {0} {0}".format("Hello", "World", 3))
```

```python
# Hello World 3
print("{h} {w} {num}".format(h="Hello", w="World", num=3))
```

```python
result = 100/777
result
```

```python
# The result was            0.1287001
print("The result was {r:20.7f}".format(r=result))
```

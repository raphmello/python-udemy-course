Inherintance

```python
class Animal:
    def __init__(self):
        print("ANIMAL CREATED")

    def who_am_i(self):
        print("I am an animal")

    def eat(self):
        print("I'm eating")

class Dog(Animal):
    def __init__(self):
        super().__init__()
        print("DOG CREATED")

    def eat(self):
        print("I am dog eating")
```

```python
my_dog = Dog()
```

```python
my_dog.who_am_i()
```

```python
my_dog.eat()
```

Polymorphism

```python
class Dog:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return self.name + ' woof'

class Cat:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return self.name + ' meow'
```

```python
niko = Dog("niko")
felix = Cat("felix")
print(niko.speak())
print(felix.speak())
```

```python
def pet_speak(pet):
    print(pet.speak())
```

```python
pet_speak(niko)
pet_speak(felix)
```

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        raise NotImplementedError("Subclass must implement speak")
```

```python
class Dog(Animal):
    def __init__(self, name):
        super().__init__(name)

    def speak(self):
        return self.name
```

```python
niko = Dog("niko")
```

```python
niko.speak()
```

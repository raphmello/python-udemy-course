```python
import math


class Dog:
    species = 'mammal'

    def __init__(self, breed, name, spots):
        self.breed = breed
        self.name = name
        self.spots = spots

    def bark(self):
        breed = self.breed.capitalize()
        print(f"I am a {breed}.")

my_dog = Dog("Lab", "Sammy", False)
```

```python
class Circle:
    pi = 3.14

    def __init__(self, radius):
        self.radius = radius

    def get_circumference(self):
        return 2 * self.pi * self.radius

my_circle = Circle(5)
```

```python
my_circle.get_circumference()
```

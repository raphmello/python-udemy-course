```python
def square(num):
    return num * num
```

```python
my_nums = [1,2,3,4,5]
```

```python
list(map(square, my_nums))
```

```python
def splicer(my_string):
    if len(my_string) % 2 == 0:
        return "EVEN"
    else:
        return my_string[0]
```

```python
name = ['Andy', 'Eve', 'Sally']
list(map(splicer, name))
```

```python
def check_even(num):
    return num % 2 == 0
```

```python
my_nums = [1,2,3,4,5,6]
```

```python
list(filter(check_even, my_nums))
```

```python
square = lambda num: num % 2 == 0
list(filter(square, my_nums))
```
